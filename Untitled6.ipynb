```python
df.columns = ['datas', 'codigo', 'tipo', 'simbolo', 'compra', 'venda', 'ns1', 'ns2']#altera o nome das colunas
df = df.rename(columns={'ns2': 'ns10'})#renomeia a coluna
df1 = df.drop(columns=['ns1', 'ns10'])#exclui a coluna
df2 = df1.drop(0)#exclui a primeira linha(linha 0)
df1.dtypes#mostra o tipo de variavel que apareceu na tabela
#compra e vendas estão errados pois no ingles eles utilizam . em vez da , para numeros e isso gera conflito, mostrando compra e venda como object(string)
#convertendo para o formato de datas
df1['datas'] = pd.to_datetime(df1['datas'], format='%d%m%Y')
#formatando para exibir dd/mm/aaaa
df1['datas'] =  df1['datas'].dt.strftime('%d/%m/%Y')
df1.head(23)
df1
df1['compra'] = df1['compra'].astype(str).apply(lambda x: x.replace(',', '.'))
df1['venda'] = df1['venda'].astype(str).apply(lambda x: x.replace(',', '.'))

df1['compra'] = df1['compra'].astype(float)
df1['venda'] = df1['venda'].astype(float)
df1.head(23)
```

```python
df1["datas"] = pd.to_datetime(df1["datas"], format= '%d/%m/%Y')#altera o formato da data
df1
```

```python
compra = df1['compra']#tabela de compras
compra
```

```python
freq_absoluta = Counter(compra)#frequencia absoluta
freq_absoluta
```

```python
tabela = pd.DataFrame.from_dict(freq_absoluta, orient='index')#transformando em dataframe
tabela
```

```python
tabela = tabela.sort_index(ascending=True)#Colocando em ordem crescente, para colocar em ordem decescente, coloca ascending = false
tabela
```

```python
#colocar os indices reais
tabela.reset_index(inplace=True)
tabela
```

```python
tabela = tabela.rename(columns={'index': 'valores', 0: 'frequencia_absoluta'})
tabela
```

```python
tabela['frequencia_rel'] = tabela['frequencia_absoluta']/tabela['frequencia_absoluta'].sum()
tabela
```

```python
tabela['frequencia_rel_perc'] = tabela['frequencia_rel']*100
tabela
```

```python
tabela['frequencia_acumulada'] = tabela['frequencia_absoluta'].cumsum()
tabela
```

```python
tabela.valores.max()
```

```python
tabela.valores.min()
```

```python
dif = tabela['valores'].max() - tabela['valores'].min()
dif
```

```python
classes = [5.3, 5.4, 5.5, 5.6, 5.7]
```

```python
labels = ['5.3-5.4','5.4-5.5','5.5-5.6','5.6-5.7']
```

```python
intervalos = pd.cut(x = tabela.valores, bins = classes, labels = labels, include_lowest= True)
```

```python
freq_absoluta = pd.value_counts(intervalos)
freq_absoluta
```

```python
frequencia_rel = pd.value_counts(intervalos, normalize = True)
frequencia_rel
```

```python
dist_freq = pd.DataFrame({'Frequencia absoluta':freq_absoluta, 'Frequencia relativa': frequencia_rel})
dist_freq
```

```python
dist_freq.sort_index(ascending=True, inplace = True)
dist_freq
```

```python
dist_freq['freq_rel_perc'] = np.round(dist_freq['Frequencia relativa percentual']*100)
dist_freq
```

```python
dist_freq['freq_acumulada'] = dist_freq['Frequencia absoluta'].cumsum()
dist_freq
```
